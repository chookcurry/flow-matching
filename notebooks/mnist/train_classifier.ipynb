```python
import torch
from diffusion.architectures.classifier.mnist_classifier import (
    SimpleCNN,
    MNISTClassifierTrainer,
)
from diffusion.data.mnist_sampleable import MNISTSampleable
```

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")
print(device)
```

```python
train_sampeable = MNISTSampleable(train=True)
val_sampeable = MNISTSampleable(train=False)

classifier = SimpleCNN(num_classes=train_sampeable.num_classes).to(device)

trainer = MNISTClassifierTrainer(
    classifier=classifier,
    train_data=train_sampeable,
    val_data=val_sampeable,
)
```

```python
trainer.train(
    num_epochs=15,
    device=device,
    lr=3e-4,
    batch_size=64,
    steps_per_epoch=500,
    validate=True,
)
```

```python
torch.save(classifier.state_dict(), "./models/classifier.pt")
```
