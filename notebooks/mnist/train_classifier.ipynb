```python
import torch
from diffusion.architectures.classifiers.mnist_classifier import MNISTClassifier
from diffusion.sampleables.mnist_sampleable import MNISTSampleable
from diffusion.training.classifier_trainer import ClassifierTrainer
from diffusion.training.encoder_trainer import EncoderTrainer
```

```python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")
print(device)
```

```python
train_sampeable = MNISTSampleable(train=True)
val_sampeable = MNISTSampleable(train=False)

classifier = MNISTClassifier(num_classes=train_sampeable.num_classes).to(device)

# trainer = ClassifierTrainer(
#     classifier=classifier,
#     train_data=train_sampeable,
#     val_data=val_sampeable,
# )

trainer = EncoderTrainer(
    encoder=classifier,
    train_data=train_sampeable,
    val_data=val_sampeable,
)
```

```python
state_dict = trainer.train(
    num_epochs=15,
    device=device,
    lr=3e-4,
    batch_size=64,
    steps_per_epoch=500,
    validate=True,
    patience=None,
)
```

```python
torch.save(classifier.state_dict(), "./models/encoder.pt")
```
